```python
import pandas as pd
import numpy as np
import os # Assuming this is imported

# -------------------------------------------------------------
# 1. HELPER FUNCTION: Calculations (Single Responsibility)
# -------------------------------------------------------------
def _calculate_metrics(team_stats_df):
    """Calculates all new rate and ratio metrics for the team stats DataFrame."""
    df = team_stats_df.copy()
    
    # Filter to only Regular Season
    df = df[df['season_type'] == 'REG']
    df = df.sort_values(by=['team', 'week'], ascending=True)
    
    attempts = df['attempts']
    total_att_carr = df['attempts'] + df['carries']
    att_sacks_suff = attempts + df['sacks_suffered']
    first_downs = df['passing_first_downs'] + df['rushing_first_downs']

    # Use a common helper for np.divide to simplify
    def safe_divide(numerator, denominator):
        return np.divide(
            numerator, 
            denominator, 
            out=np.zeros_like(numerator, dtype=float), 
            where=denominator != 0
        )

    # Calculate Rates
    df['pass_rate']       = safe_divide(attempts, total_att_carr)
    df['yards_per_att']   = safe_divide(df['passing_yards'], attempts)
    df['td_rate']         = safe_divide(df['passing_tds'], attempts)
    df['int_rate']        = safe_divide(df['passing_interceptions'], attempts)
    df['sack_rate']       = safe_divide(df['sacks_suffered'], att_sacks_suff)
    df['rush_ypc']        = safe_divide(df['rushing_yards'], df['carries'])
    df['explosive_proxy'] = safe_divide(first_downs, total_att_carr)
    df['adot']            = safe_divide(df['passing_air_yards'], attempts)
    df['yac_per_rec']     = safe_divide(df['passing_yards_after_catch'], df['receptions'])
    
    # Calculate EPA metrics
    df['off_epa'] = df['passing_epa'] + df['rushing_epa'] + df['receiving_epa']
    df['epa_per_play'] = safe_divide(df['off_epa'], total_att_carr)
    
    return df

# -------------------------------------------------------------
# 2. MAIN FUNCTION: Orchestration (Handles Flow Control and I/O)
# -------------------------------------------------------------
def process_all_years(years, base_path='data/raw'):
    """
    Loops through years, processes team stats, merges with odds data, 
    and saves the resulting DataFrame to a year-specific Parquet file.
    """
    
    # Load static data ONCE outside the loop (for efficiency)
    try:
        odds_data_names_norm = pd.read_parquet(os.path.join(base_path, 'odds_names_norm.parquet'))
    except Exception as e:
        print(f"Error loading required odds data: {e}")
        return # Exit if a crucial file can't be loaded

    # TEAM_TO_ABBR is defined but not used in the final code logic, 
    # so it can be omitted or moved outside if it's not needed for processing.
    
    for year in years:
        print(f"--- Processing year: {year} ---")
        
        # 1. Setup paths
        season_folder = os.path.join(base_path, str(year))
        os.makedirs(season_folder, exist_ok=True)
        
        # 2. Load year-specific data
        try:
            team_stats_path = os.path.join(base_path, str(year), 'team_stats.parquet')
            team_stats = pd.read_parquet(team_stats_path)
        except Exception as e:
            print(f"Skipping {year}: Could not load team_stats.parquet. Error: {e}")
            continue # Move to the next year
            
        # 3. Process the data (calling the helper function)
        team_stats_processed = _calculate_metrics(team_stats)
        
        # 4. Merge DataFrames
        merged_df = team_stats_processed.merge(
            odds_data_names_norm,
            on=['season', 'team', 'week'],
            how='left'
        )

        # 5. Save the final Parquet file
        file_name = f'{year}_weekly_offense_and_odds.parquet'
        file_path = os.path.join(season_folder, file_name)
        
        try:
            merged_df.to_parquet(file_path, index=False) # Recommend index=False unless index is meaningful
            print(f"SUCCESS: Saved {len(merged_df)} rows to {file_path}")
        except Exception as e:
            print(f"FAILURE: Could not save data for {year}. Error: {e}")

    print("--- Processing complete for all specified years. ---")
```

```python
import pandas as pd
import os
import glob # Useful for finding files

def calculate_and_save_defense(years, base_path='data/raw'):
    """
    Loops through each year, loads the offense data, transforms it into
    defensive metrics (by renaming columns), and saves the resulting
    defensive DataFrame as a new Parquet file.
    """
    
    # 1. Define the necessary column renames and final column list
    # The team and opponent columns are swapped to shift perspective from offense to defense.
    RENAME_MAP = {
        "opponent_team": "team",         # The opponent team becomes the 'defensive team'
        "team": "off_team",              # The original team becomes the 'offensive team'
        "loc": "off_team_loc",
        
        # Offensive stats become defensive stats allowed
        "passing_yards": "passing_yards_allowed",
        "passing_air_yards": "passing_air_yards_allowed",
        "passing_yards_after_catch": "passing_yards_after_catch_allowed",
        "passing_tds": "passing_tds_allowed",
        "rushing_yards": "rushing_yards_allowed",
        "rushing_tds": "rushing_tds_allowed",
        "receiving_yards": "receiving_yards_allowed",
        "receiving_tds": "receiving_tds_allowed",
        
        # Rate stats become allowed/forced rates
        "off_epa": "def_epa_allowed",
        "epa_per_play": "epa_per_play_allowed",
        "pass_rate": "pass_rate_allowed",
        "yards_per_att": "yards_per_att_allowed",
        "rush_ypc": "rush_ypc_allowed",
        "explosive_proxy": "explosive_allowed",
        "td_rate": "td_rate_allowed",
        "int_rate": "int_rate_forced",  # Renamed to reflect defense forced it
        "sack_rate": "sack_rate_forced", # Renamed to reflect defense forced it
        "adot": "adot_allowed",
        "yac_per_rec": "yac_per_rec_allowed",
    }
    
    # These are the columns to keep and save in the final defensive DataFrame
    FINAL_DEFENSE_COLUMNS = [
        "season", "week", "team", "off_team", "off_team_loc",
        
        # Metrics derived from offense stats
        "passing_yards_allowed", "passing_air_yards_allowed",
        "passing_yards_after_catch_allowed", "passing_tds_allowed",
        "rushing_yards_allowed", "rushing_tds_allowed",
        "receiving_yards_allowed", "receiving_tds_allowed",
        "def_epa_allowed", "epa_per_play_allowed",
        "pass_rate_allowed", "yards_per_att_allowed",
        "rush_ypc_allowed", "explosive_allowed",
        "td_rate_allowed", "int_rate_forced",
        "sack_rate_forced", "adot_allowed",
        "yac_per_rec_allowed",

        # Existing defensive stats that need to be carried over from the original frame
        # NOTE: If these columns (def_sacks, etc.) are NOT in the offense file, 
        # they must be merged in separately from another source or removed from this list.
        "def_sacks", "def_qb_hits", "def_interceptions",
        "def_fumbles_forced", "def_pass_defended",
        "def_tds", "def_safeties", 
    ]


    for year in years:
        print(f"--- Starting defensive processing for {year} ---")
        
        # 1. Define input and output paths
        season_folder = os.path.join(base_path, str(year))
        
        input_file = os.path.join(season_folder, f'{year}_weekly_offense_and_odds.parquet')
        output_file = os.path.join(season_folder, f'{year}_weekly_defense.parquet')
        
        # 2. Check if input file exists
        if not os.path.exists(input_file):
            print(f"File not found: {input_file}. Skipping {year}.")
            continue
            
        # 3. Load the offensive data
        try:
            weekly_offense = pd.read_parquet(input_file)
        except Exception as e:
            print(f"Error loading offense data for {year}: {e}. Skipping.")
            continue
            
        # 4. Create the defensive DataFrame (def_df) by renaming columns
        # This renames columns to shift the perspective from the scoring team to the defending team.
        def_df = weekly_offense.rename(columns=RENAME_MAP)
        
        # 5. Merge Step: Re-add the original team column (now 'off_team') 
        # This step seems redundant given the RENAME_MAP, but it's kept to align with your original logic
        # (It seems you were trying to ensure the offense team name was correctly linked back)
        
        # The key logic here is that for every row in def_df, the "team" (now defense) 
        # must match the "opponent_team" in the original weekly_offense.
        # Since the columns are already renamed in def_df, we just ensure we keep the necessary columns.
        
        # The merge operation you used previously:
        # def_df = def_df.merge(weekly_offense[["season","week","opponent_team"]],
        #                       left_on=["season","week","team"],
        #                       right_on=["season","week","opponent_team"],
        #                       how="left")
        
        # The merge is not necessary here because all required data (the opponent team, now named 'off_team') 
        # is already present and renamed correctly in 'def_df'. 
        
        # 6. Filter to the final, clean set of defensive columns
        # Note: We must check which columns actually exist in the DataFrame after renaming/merging.
        
        # Create a list of columns that are guaranteed to exist after the RENAME_MAP is applied
        existing_cols = list(def_df.columns)
        
        # Filter FINAL_DEFENSE_COLUMNS to only include those present in the DataFrame
        final_cols = [col for col in FINAL_DEFENSE_COLUMNS if col in existing_cols]
        
        # Check for missing defensive stats like 'def_sacks'
        missing_cols = [col for col in FINAL_DEFENSE_COLUMNS if col not in existing_cols and col.startswith('def_')]
        if missing_cols:
            print(f"Warning: Missing core defensive stats in source data for {year}: {missing_cols}")


        def_df = def_df[final_cols]
        
        # 7. Save the final Parquet file
        try:
            def_df.to_parquet(output_file, index=False)
            print(f"SUCCESS: Saved {len(def_df)} rows to {output_file}")
        except Exception as e:
            print(f"FAILURE: Could not save defensive data for {year}. Error: {e}")
            
    print("--- Defensive processing complete for all specified years. ---")

# Example usage (assuming you have the necessary files):
# if __name__ == '__main__':
#     # NOTE: The base_path needs to be the same as the one used in the offense creation
#     calculate_and_save_defense(years=[2022, 2023, 2024], base_path='data/raw')

# calculate_and_save_defense(years=list(range(2016,2025)))
```
